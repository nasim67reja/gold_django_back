```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('capital_curve_data.csv')

# df["opening_time_bd"] = pd.to_datetime(df["opening_time_bd"])
# df = df.sort_values("opening_time_bd").reset_index(drop=True)

df
```

```python
df["opening_time_bd"] = pd.to_datetime(df["opening_time_bd"])
df = df.sort_values("opening_time_bd").reset_index(drop=True)
df
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df["opening_time_bd"], df["capital"])
plt.xlabel("Time (Bangladesh)")
plt.ylabel("Capital (USD)")
plt.title("Equity Curve")
plt.grid(True)
plt.show()
```
