```python
import pandas as pd

df = pd.read_csv('filtered_trade_history_update.csv')


df["opening_time_bd"] = pd.to_datetime(df["opening_time_bd"])


end_date = pd.Timestamp("2026-01-06 23:59:59")

filtered_df = df[df["opening_time_bd"] <= end_date]

filtered_df.tail()
```

```python
# Make sure numeric (very important)
filtered_df["profit_usd"] = pd.to_numeric(
    filtered_df["profit_usd"], errors="coerce").fillna(0)
filtered_df["commission_usd"] = pd.to_numeric(
    filtered_df["commission_usd"], errors="coerce").fillna(0)

# Starting capital
starting_capital = 103.67


# Capital curve (running total)
filtered_df["capital"] = starting_capital + filtered_df["trade_pnl"].cumsum()

filtered_df.to_csv('capital_curve_data.csv', index=False)
```

```python
df = pd.read_csv("filtered_trade_history_update.csv")

df["opening_time_bd"] = pd.to_datetime(df["opening_time_bd"])
df = df.sort_values("opening_time_bd").reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(df["opening_time_bd"], df["capital"])
plt.xlabel("Time (Bangladesh)")
plt.ylabel("Capital (USD)")
plt.title("Equity Curve")
plt.grid(True)
plt.show()
```
