```python
import yfinance as yf
import pandas as pd

df = yf.download(
    tickers="GC=F",
    interval="1h",
    start="2025-01-01",
    end="2026-01-18",
    progress=False
)
df.reset_index(inplace=True)

# flatten multi-index columns
df.columns = df.columns.get_level_values(0)


# convert utc to bd

df["Datetime"] = pd.to_datetime(df["Datetime"], utc=True)
df["Datetime_bd"] = df["Datetime"].dt.tz_convert("Asia/Dhaka")


# add candle direction
df["candle_body"] = (df["Close"] - df["Open"]) / 0.01

# add price movement
df["price_movement_pips"] = (df["High"] - df["Low"]) / 0.01

# df["Datetime"] = pd.to_datetime(df["Datetime"])
df["hour"] = df["Datetime_bd"].dt.hour



# df.tail(20)
```

## Create new DataFrame with hourly total movement

```python
hourly_movement_df = (
    df.groupby("hour", as_index=False)
      .agg(total_movement_pips=("price_movement_pips", "sum"))
)

hourly_movement_df = hourly_movement_df.sort_values("total_movement_pips")

hourly_movement_df
```
